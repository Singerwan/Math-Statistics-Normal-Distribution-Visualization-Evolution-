```python
x=np.linspace(-4,4,51)
e=sp.N(sp.E)
y1=1 / (  e** ( x**2 ))
y2=  e** - ( x**2 )

fig,ax=plt.subplots(1,2,figsize=(12,3))

ax[0].plot(x,y1,ls="--",marker="*",alpha=0.4,c="red")
ax[0].axvline(0,0,1,ls="--",c="grey",alpha=0.3)
ax[0].axvline(x[20],0,0.51,ls="--",c="blue",alpha=0.3)
ax[0].axhline(y1[20],0,0.4,ls="--",c="blue",alpha=0.3)

ax[0].text(s="f(x)=$\\frac{1}{e^{x^2}}$",x=-4.2,y=0.8,size=16,color="red")
ax[0].text(s="As x in creases, \n y decreases",x=0.8,y=0.7,size=16,color="red")

ax[1].text(s="f(x)=${e^{-(x^2)}}$",x=-4.2,y=0.8,size=16,color="blue")
ax[1].text(s="$f(x) = \\frac{1}{\\sigma \\  \\sqrt{2 \\cdot \\pi}}  e^{ -\\frac{(x - \\mu)^2}{2 \\cdot \\sigma^2} }$",x=0.58,y=0.8,size=14,color="blue")
ax[1].text(s="$\\sigma$ spread stretch\n $\\mu$ center shift ",x=1,y=0.4,size=12,color="green")
ax[1].text(s="area under \ncurve is $\\sqrt{\\pi}$" ,x=-1,y=0.2)


ax[1].plot(x,y2)
plt.suptitle("Fundamental Formula for Normal Distribution",size=20,color="indigo")

plt.show()
```
